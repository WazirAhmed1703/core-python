```python
#Temperature Converter: Write a program that converts Celsius to Fahrenheit and vice versa based on user choice.
convertion_type = int(input("Enter 1 for Celsius to Fahrenheit and 2 for Fahrenheit to Celsius: "))
# convertion_type = 2

match convertion_type:
    case 1:
        degree_celsius = float(input("Enter temperature in Celsius: "))
        # degree_celsius = 31.2
        result = degree_celsius * (9/5) + 32
        print(result)
    case 2:
        degree_fahrenheit = float(input("Enter temperature in Fahrenheit: "))
        # degree_fahrenheit = 88.16
        result = (degree_fahrenheit - 32) * (5/9)
        print(result)
    case _:
        print("Please select right option!!!")
```
