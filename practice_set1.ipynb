```python
#Temperature Converter: Write a program that converts Celsius to Fahrenheit and vice versa based on user choice.
convertion_type = int(input("Enter 1 for Celsius to Fahrenheit and 2 for Fahrenheit to Celsius: "))
# convertion_type = 2

match convertion_type:
    case 1:
        degree_celsius = float(input("Enter temperature in Celsius: "))
        # degree_celsius = 31.2
        result = degree_celsius * (9/5) + 32
        print(result)
    case 2:
        degree_fahrenheit = float(input("Enter temperature in Fahrenheit: "))
        # degree_fahrenheit = 88.16
        result = (degree_fahrenheit - 32) * (5/9)
        print(result)
    case _:
        print("Please select right option!!!")
```

```python
#Leap Year Validator: Check whether a given year is a leap year (divisible by 4, except centuries unless divisible by 400).

year = int(input("Enter a year: "))
if(year % 4 == 0 and year % 100 != 0):
    print(f"{year} is Leap Year!!!")
elif(year % 400 == 0):
    print(f"{year} is a Leap Year!!!")
else:
    print(f"{year} is not a Leap Year!!!")
```

```python
#Grade Calculator: Accept marks for 5 subjects, compute the percentage, and assign grades (A, B, C, D, or F) using if-elif-else.

print("Welcome to Wazir's Classroom!!!")
print("Enter five subjects score: ")
marks = []
for i in range(1,6):
   marks.append(int(input(f"Enter subject No:{i}:")))
result=sum(marks)
avg = result // 5
if(avg>=90):
    print('A')
elif(avg >= 70 and avg < 90):
    print('B')
elif(avg >= 50 and avg < 70):
    print('C')
else:
    print('F')
```
