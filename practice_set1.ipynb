```python
#Leap Year Validator: Check whether a given year is a leap year (divisible by 4, except centuries unless divisible by 400).

year = int(input("Enter a year: "))
if(year % 4 == 0 and year % 100 != 0):
    print(f"{year} is Leap Year!!!")
elif(year % 400 == 0):
    print(f"{year} is a Leap Year!!!")
else:
    print(f"{year} is not a Leap Year!!!")
```

```python
# Factorial Calculator: Calculate the factorial of a non-negative integer using a for loop.

number = int(input("Enter a number to get factorial:"))

def fact(num):
    if num == 1 or num == 0:
        return 1
    return num* fact(num - 1)

print(f"Factorial: {fact(number)}")
```

```python
#Palindrome Check: Check if a string reads the same forward and backward, ignoring spaces and case.

word = str(input("Enter a word:"))

rev = word[::-1]

if(rev == word):
    print(f"{word} is a Palindrome")
else:
    print(f"{word} is not a Palindrome")
```

```python
#Vowel & Consonant Counter: Count the total number of vowels, consonants, digits, and spaces in a string.

sentence = str(input("Enter a sentence:"))
vowels = 0
constonants = 0
digits = 0
spaces = 0

for i in sentence:
    if i.isalpha():
        if i in ['a','i','o','u','e']:
            vowels += 1
        else:
            constonants += 1
    elif i.isspace():
        spaces += 1
    elif i.isdigit():
        digits += 1

print(f"Vowels: {vowels}, Consonants: {constonants}, Digits: {digits}, Spaces: {spaces}")
```

```python
#Anagram Detector: Check if two strings are anagrams of each other (e.g., "listen" and "silent").
str_one = str(input("Enter a string one :"))
str_two = str(input("Enter a string two :"))

str_one = sorted(str_one)
str_two = sorted(str_two)

if str_one == str_two:
    print("Anagrams")
else:
    print("Non Anagrams")
```
