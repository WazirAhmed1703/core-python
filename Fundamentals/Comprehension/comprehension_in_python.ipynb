# Comprehension
Comprehension are a concise way to create lists,sets,dictionaries or generators in Python using a single lines of code.

we use it because it is cleaner code and Faster execution.

#### List Comprehension

[expression for item in iterable if condition]


```python
list_of_item = [
    "Veg Biryani",
    "Mutton Biryani",
    "Chicken Biryani",
    "Veg Tikka",
    "Veg Butter Kabab"
]

veg_item = [item for item in list_of_item if "Veg" in item]

veg_item
```

#### Set Comprehension

{expression for item in iterator if condition}

```python
players_favrate_team = {
    "kolkata":["Mohammad Shami","Mohammad Siraj","Rohit","Virat"],
    "chennai":["Ravi teja","Anna Reddy","Mohammad Shami","Dhoni","Virat"]
}

list_of_unique_player = {player for players in players_favrate_team.values() for player in players}

list_of_unique_player
```

#### Dictionary Comprehension

{expression for item in iterator if condition}

```python
cloth_price_inr = {
    "shirt":500,
    "t-shirt":450,
    "jeans":1000,
    "blanket":1500
}

cloth_price_usd = { item:price/80 for item,price in cloth_price_inr.items()}

cloth_price_usd
```

#### Generator

(expression for item in iterable if condition)

[x for x in items] --> make entire lists in memory

(x for x in items) --> like a stream 

```python
total_cash_collected = [2500,55600,4478,115,88095,1445,2202,2203,55689]

total_cash = sum (x for x in total_cash_collected if x > 10000)

total_cash
```

### Practice Set

```python
# Even or Odd Labels: Given nums = [1, 2, 3, 4, 5], create a list of strings: ["Odd", "Even", "Odd", "Even", "Odd"].
nums = [1, 2, 3, 4, 5]

odd_or_even = ["Even" if num % 2 == 0 else "Odd" for num in nums]

odd_or_even
```

```python
# Divisible by 3 and 5: From range(1, 101), find all numbers divisible by both 3 and 5.

three_and_five = [num for num in range(1,101) if num%3==0 and num%5==0]

three_and_five
```

```python
# Scenario: You are parsing server logs and need to extract all critical error messages while ignoring timestamps and info logs.

logs = [
    "2026-11-20 10:01:12 INFO User logged in",
    "2026-11-20 10:02:45 ERROR Database connection failed",
    "2026-11-20 10:03:10 WARNING High memory usage",
    "2026-11-20 10:05:00 ERROR Timeout on payment gateway"
]

list_error_message = list(log.split("ERROR")[1].strip() for log in logs if "ERROR" in log)  

list_error_message
```

```python
# A user uploaded a list of raw email addresses with inconsistent whitespace, mixed casing, and empty entries.
raw_emails = ["  Alice@Example.com ", "", "bob@domain.com", "   ", "CHARLIE@Site.ORG\n"]

list_of_email = list(email.strip() for email in raw_emails if email.strip() != "")

list_of_email
```

```python
# Scenario: You need to apply a 10% discount to all in-stock items priced over ₹1,000, while keeping other in-stock items at their original price. Out-of-stock items should be excluded.

carts = [
    {"item": "Mechanical Keyboard", "price": 4500, "in_stock": True},
    {"item": "USB-C Cable", "price": 400, "in_stock": True},
    {"item": "27-inch Monitor", "price": 18000, "in_stock": False},
    {"item": "Wireless Mouse", "price": 1500, "in_stock": True}
]

discount_item = [

    (cart["item"],cart["price"] * 0.9  if cart["price"] > 1000 else cart["price"])
    for cart in carts
    if cart["in_stock"] == True
]

discount_item
```

```python
# Scenario: An API returns a list of departments, each containing a list of employee records. You need a flat list of emails for active engineers only.
company_data = [
    {
        "dept": "Engineering",
        "employees": [
            {"name": "Aarav", "email": "aarav@corp.com", "active": True},
            {"name": "Neha", "email": "neha@corp.com", "active": False}
        ]
    },
    {
        "dept": "Marketing",
        "employees": [
            {"name": "Rohan", "email": "rohan@corp.com", "active": True}
        ]
    }
]

active_engineer = [
        employee["email"] for department in company_data if department["dept"] == "Engineering"
        for employee in department["employees"] if employee["active"]
    ]

print(active_engineer)
```
