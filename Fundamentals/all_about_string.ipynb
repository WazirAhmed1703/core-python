# String Methods

```python
string = "This is a sentence that will help us to understand String in Python buildin methods."

# --- Case Conversion ---
string.capitalize() # to change the first character to uppercase
string.casefold() # to convert string into lower case (aggressive, for caseless matching)
string.lower() # to convert a string into lower case
string.swapcase() # to swap cases, lower case becomes upper case and vice versa
string.title() # to change the first character of each word to uppercase
string.upper() # to convert a string into upper case

# --- Search & Count ---
string.count('t') # to return the number of times a specified value occurs in a string
string.endswith('.') # to return True if the string ends with the specified value
string.find('help') # to search the string for a specified value and return its position (-1 if not found)
string.index('help') # to search the string for a specified value and return its position (raises error if not found)
string.rfind('t') # to search from the right and return the last position of where it was found (-1 if not found)
string.rindex('t') # to search from the right and return the last position of where it was found (raises error if not found)
string.startswith('This') # to return True if the string starts with the specified value

# --- Modification & Formatting ---
string.center(100, '-') # to return a centered string padded with a specified character
string.encode() # to return an encoded version of the string (default UTF-8)
string.expandtabs(4) # to set the tab size of the string
"Hello {}".format("World") # to format specified values in a string
"{name}".format_map({'name': 'Python'}) # to format specified values in a string using a dictionary
string.ljust(100, '-') # to return a left-justified version of the string
string.lstrip() # to remove leading whitespace (or specified characters) from the left
string.removeprefix('This ') # to remove the specified prefix from the start of the string (Python 3.9+)
string.removesuffix('.') # to remove the specified suffix from the end of the string (Python 3.9+)
string.replace('Python', 'Java') # to return a string where a specified value is replaced with another
string.rjust(100, '-') # to return a right-justified version of the string
string.rstrip() # to remove trailing whitespace (or specified characters) from the right
string.strip() # to remove both leading and trailing whitespace (or specified characters)
string.zfill(100) # to fill the string with a specified number of 0 values at the beginning

# --- Splitting & Joining ---
"-".join(['a', 'b', 'c']) # to join the elements of an iterable to the end of the string
string.partition('help') # to split the string into a 3-element tuple (before, match, after)
string.rpartition('t') # to split from the right into a 3-element tuple (before, match, after)
string.rsplit(' ', 2) # to split the string from the right at the specified separator and return a list
string.split(' ') # to split the string at the specified separator and return a list
string.splitlines() # to split the string at line breaks and return a list

# --- Translation ---
table = str.maketrans('a', '@') # to return a translation table to be used in translations
string.translate(table) # to return a translated string using the mapping table

# --- Validation & Checks (Returns True or False) ---
string.isalnum() # to return True if all characters in the string are alphanumeric
string.isalpha() # to return True if all characters in the string are in the alphabet
string.isascii() # to return True if all characters in the string are ASCII characters
string.isdecimal() # to return True if all characters in the string are decimals (0-9)
string.isdigit() # to return True if all characters in the string are digits
string.isidentifier() # to return True if the string is a valid Python identifier/variable name
string.islower() # to return True if all characters in the string are lower case
string.isnumeric() # to return True if all characters in the string are numeric
string.isprintable() # to return True if all characters in the string are printable
string.isspace() # to return True if all characters in the string are whitespaces
string.istitle() # to return True if the string follows the rules of a title
string.isupper() # to return True if all characters in the string are upper case

```

```python
# Exercise 1: Clean & Format User Input
# Scenario: A user entered their name with messy spacing and capitalization.
# Goal: Clean it up so it outputs "John Doe"

username = input("Enter your name: ")
username = username.strip().title()
print(username)
```

```python
# Exercise 2: Extract Domain from an Email
# Scenario: You need to extract the domain name ("gmail.com") from an email address.

email = input("Enter your email id: ")
list_of = email.split('@')
print(list_of[1])
```

```python
# Exercise 3: Check File Extension
# Scenario: Verify if an uploaded file is a PDF (case-insensitive, so ".PDF" should also pass).
filename = "Project_Report_2026.PDF"

is_pdf = filename.lower().endswith('.pdf')
is_pdf 
```

```python
# Exercise 4: Format an Invoice ID
# Scenario: Database IDs must be 6 digits long, padded with leading zeros (e.g., "000042").
order_id = "42"

formatted_id = order_id.zfill(6)
formatted_id
```

```python
# Exercise 5: Mask a Credit Card Number
# Scenario: Show only the last 4 digits of a card number, replacing the rest with asterisks: "************3456"
card_number = "1234567890123456"

masked_card =  card_number[-4:].rjust(16,'*')
masked_card
```

```python
# Exercise 6: Clean Up a Messy URL Slug
# Scenario: Convert a blog title into a URL-friendly slug: "10-tips-for-python-beginners"
blog_title = "  10 Tips for Python Beginners!  "

slug = blog_title.strip().lower().removesuffix('!').replace(' ','-')
slug
```

```python
# Exercise 7: Validate a PIN Code
# Scenario: Check if a user's ATM PIN is strictly 4 digits (no letters, spaces, or decimals).
pin = "1234"

is_valid_pin = len(pin)==4 and pin.isdigit()
is_valid_pin
```
