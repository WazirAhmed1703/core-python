```python
#Constructor and instance method
class teacher:
    school_name = "St.John's High School"
    def __init__(self,tname,tage,tsubject): #Constructor
        self.tname = tname
        self.tage = tage
        self.tsubject = tsubject
    
    def print_details(self): #instance Method
        return f"Hi my name is {self.tname}, my age is {self.tage} and i love to tought {self.tsubject} subject"

tobj = teacher("Sahil",26,"Python")    

print(tobj.print_details())
```

```python
#inheritance
class student(teacher): #Single Inheritance
    def __init__(self,sname,sage,ssubject,tname,tage,tsubject):
        self.sname = sname
        self.sage = sage
        self.ssubject = ssubject
        super().__init__(tname,tage,tsubject)
    
    def print_student_details(self):
        teacher_details=super().print_details()
        return f"{teacher_details}\nSchool Name: {self.school_name}\nStudent name: {self.sname}\nStudent age: {self.sage}\nSubject:{self.ssubject}"
    
std1 = student("Wazir",26,"Python","Sahil",26,"Data Engineer")   

print(std1.print_student_details())
```

```python
#Multilevel inheritance
class School:
    def __init__(self, school_name):
        self.school_name = school_name

    def print_details(self):
        return f"School: {self.school_name}"


class Teacher(School):
    def __init__(self, school_name, tname, tage, tsubject):
        super().__init__(school_name)
        self.tname = tname
        self.tage = tage
        self.tsubject = tsubject

    def print_details(self):
        school_details = super().print_details()
        return (
            f"{school_details}\n"
            f"Teacher: {self.tname}\n"
            f"Teacher age: {self.tage}\n"
            f"Teacher subject: {self.tsubject}"
        )

class Student(Teacher):
    def __init__(
        self, school_name, tname, tage, tsubject,
        sname, sage, ssubject
    ):
        super().__init__(school_name, tname, tage, tsubject)
        self.sname = sname
        self.sage = sage
        self.ssubject = ssubject

    def print_student_details(self):
        teacher_details = super().print_details()
        return (
            f"{teacher_details}\n"
            f"Student: {self.sname}\n"
            f"Student age: {self.sage}\n"
            f"Student subject: {self.ssubject}"
        )


std1 = Student(
    "St. John's High School",
    "Sahil", 26, "Data Engineering",
    "Wazir", 26, "Python"
)

print(std1.print_student_details())
        
```

```python
#Multiple Inheritance 
class company:
    __company_name = "Bashir Ahmed and son's" # private
    _ceo = "Wazir Ahmed" # protected 

    def __init__(self,business_type):
        self.business_type = business_type
    @classmethod
    def get_company_name(cls):
        return cls.__company_name
    @classmethod
    def get_ceo(cls):
        return cls._ceo
    
    def print_company_details(self):
        return f"Type of Business: {self.business_type}"

class vitual_mba:
    __tution_name = "Bits Pilani"
    _course_type = "Human Resource"
    
    def __init__(self,tution_fee):
        self.tution_fee = tution_fee
    @classmethod
    def get_tution_name(cls):
        return cls.__tution_name
    @classmethod
    def get_course_type(cls):
        return cls._course_type

    def print_tution_details(self):
        return f"Tution Fee: {self.tution_fee}"

class employee(company,vitual_mba):
    def __init__(self, business_type, tution_fee):
        company.__init__(self, business_type)
        vitual_mba.__init__(self, tution_fee)
    
    def printDetails(self):
        com_name = company.get_company_name()
        com_ceo = company.get_ceo()
        tut_name = vitual_mba.get_tution_name()
        tut_type = vitual_mba.get_course_type()
        busi_details = super().print_company_details()
        tut_details = super().print_tution_details()
        return f"{com_name}:{com_ceo}\n{tut_name}:{tut_type}\n{busi_details}:{tut_details}"

emp = employee("Leather",500000)
 
print(emp.printDetails())
```
