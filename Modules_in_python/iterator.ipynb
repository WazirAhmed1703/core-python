```python
from itertools import groupby
```

```python
data = [
    {'dept':'IT','amount':350},
    {'dept':'Audit','amount':200},
    {'dept':'HR','amount':300},
    {'dept':'Audit','amount':700},
    {'dept':'IT','amount':1000},
    {'dept':'HR','amount':600}
]

sort_with_dept = sorted(data,key=lambda x: x['dept'])

sort_with_dept
```

```python

for dept, group in groupby(sort_with_dept, key=lambda row: row['dept']):
    total = sum(i['amount'] for i in group)
    print(dept, total)

# Note: 
# 1. It should be sorted
# 2. It return itertools.groupby iterator
# 3. we need to group
```

```python
list_of_fruit =["apple", "apricot", "banana", "blueberry", "cherry"]

for key,group in groupby(list_of_fruit,lambda x: x[0]):
    print(key,list(group))
```

```python
from itertools import count

for num in count(start=1,step=2):
    if(num > 5):
        break
    print(num)
```

```python
from itertools import cycle

rainbow = ['Violet','Blue','Green','Yellow','Orange','Red']

count = 0
for color in cycle(rainbow):
    print(color)
    count+=1
    if count == 12:
        break
```

```python
from itertools import repeat
for msg in repeat('Loading....',8):
    print(msg)
```

```python
from itertools import permutations,combinations
```

```python
teams = ['IND','AUS','WI','NZ','BNG']

for p in permutations(teams,2):
    print(p)
```

```python
for c in combinations(teams,2):
    print(c)
```

```python
from itertools import islice

data=iter(range(30))
for i in islice(data,7,11):
    print(i)
```

```python
from itertools import tee

numbers = (x*5 for x in range(1,5))

its,its2,its5 = tee(numbers,3)

print(list(its))
print(list(its2))
print(list(its5))
```

```python
from itertools import product

colors = ["Red", "Blue"]
sizes = ["S", "M", "L"]

combos = list(product(colors, sizes))
print(combos)
```
